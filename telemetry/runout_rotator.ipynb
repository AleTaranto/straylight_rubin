```python
import matplotlib.pyplot as plt
import astropy.units as u
import numpy as np
import lsst.daf.butler as DafButler
import astropy.coordinates
```

```python
butler = DafButler.Butler('embargo', collections=["LSSTCam/runs/nightlyValidation",'u/ataranto/lsst_outputs'], instrument='LSSTCam')
registry = butler.registry

day_obs=20250528
clock=[538,539,540,541,542,543,544]
clock2=[555,556,557,558,559,560,561,562,563]
```

```python
from astropy.coordinates import SkyCoord

# Collect RA/Dec for all exposures
ra_list = []
dec_list = []
ra_list2 = []
dec_list2 = []

for exp in clock:
    visit = day_obs * 100000 + exp
    img = butler.get("post_isr_image", exposure=visit, detector=94)
    wcs = img.wcs
    ra=(wcs.getSkyOrigin().getRa().asDegrees())
    dec=(wcs.getSkyOrigin().getDec().asDegrees())
    print(ra,dec)
    ra_list.append(ra)
    dec_list.append(dec)

# Compute angular distances in arcseconds from the first point
coords = SkyCoord(ra=ra_list*u.deg, dec=dec_list*u.deg)
sep_arcsec = coords[0].separation(coords).arcsec

# Scatter plot
# Compute offsets in arcseconds from the first point
ra0 = ra_list[0]
dec0 = dec_list[0]
# Convert RA offset to arcsec, accounting for declination
delta_ra_arcsec = (np.array(ra_list) - ra0) * 3600 * np.cos(np.deg2rad(dec0))
delta_dec_arcsec = (np.array(dec_list) - dec0) * 3600

plt.figure(figsize=(8,6))
plt.scatter(delta_ra_arcsec[1:], delta_dec_arcsec[1:], c='blue', marker='o', label='Other points')
plt.scatter(0, 0, c='red', marker='*', s=200, label='First point')
plt.xlabel('ΔRA (arcsec)')
plt.ylabel('ΔDec (arcsec)')
plt.title('Offsets from First Visit (arcsec)')
plt.legend()
plt.grid(True)
plt.show()


for exp in clock2:
    visit = day_obs * 100000 + exp
    img = butler.get("post_isr_image", exposure=visit, detector=94)
    wcs = img.wcs
    ra=(wcs.getSkyOrigin().getRa().asDegrees())
    dec=(wcs.getSkyOrigin().getDec().asDegrees())
    print(ra,dec)
    ra_list2.append(ra)
    dec_list2.append(dec)

# Compute angular distances in arcseconds from the first point
coords2 = SkyCoord(ra=ra_list2*u.deg, dec=dec_list2*u.deg)
sep_arcsec2 = coords2[0].separation(coords2).arcsec

ra1 = ra_list2[0]
dec1 = dec_list2[0]
# Convert RA offset to arcsec, accounting for declination
delta_ra_arcsec2 = (np.array(ra_list2) - ra1) * 3600 * np.cos(np.deg2rad(dec1))
delta_dec_arcsec2 = (np.array(dec_list2) - dec1) * 3600

plt.figure(figsize=(8,6))
plt.scatter(delta_ra_arcsec2[1:], delta_dec_arcsec2[1:], c='blue', marker='o', label='Other points')
plt.scatter(0, 0, c='red', marker='*', s=200, label='First point')
plt.xlabel('ΔRA (arcsec)')
plt.ylabel('ΔDec (arcsec)')
plt.title('Offsets from First Visit (arcsec)')
plt.legend()
plt.grid(True)
plt.show()

```
